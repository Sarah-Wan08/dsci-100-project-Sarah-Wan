```r
num_obs <- nrow(madness_tidy)

bid_proportions <- madness_tidy |>
    group_by(bid) |>
    summarize(count = n(),
              percentage = n() / num_obs * 100)

bid_bar <- bid_proportions |>
    ggplot(aes(x = bid, y = percentage)) +
    geom_bar(stat = "identity") +
    labs(x = "Bid Type",
         y = "Percentage of Teams (%)",
         title = "Proportion of Teams by Tournament Bid Type") +
    theme(text = element_text(size = 12))
bid_bar
```

Visualization 1: This bar chart shows the average number of tournament wins for each seed. Lower seeds tend to win noticeably more games, while seeds 9–16 rarely win more than one game. This suggests that seed is a strong predictor of tourney_wins and supports including it in our model. The relationship looks non-linear; wins drop sharply from seed 1 to 4, then flatten, which is one reason KNN regression may be more suitable than linear regression.

```r
seed_summary <- madness_tidy |>
    group_by(seed) |>
    summarize(mean_tourney_wins = mean(tourney_wins))

seed_plot <- seed_summary |>
    ggplot(aes(x = seed, y = mean_tourney_wins)) +
    geom_bar(stat = "identity") +
    labs(x = "Tournament Seed (1 = best)",
         y = "Average Number of Tournament Wins",
         title = "Average Tournament Wins by Seed") +
    theme(text = element_text(size = 12))
seed_plot
```

Visualization 2: This scatter plot shows a positive relationship between regular-season win percentage and tournament wins: teams that won more often during the season tend to go further in the tournament. The relationship is weaker and more spread out than with seed, but teams with high win percentages are the main ones reaching 4+ wins, so reg_wins_pct still adds useful information as a predictor.

```r
pct_plot <- madness_tidy |>
    ggplot(aes(x = reg_wins_pct, y = tourney_wins)) +
    geom_point(alpha = 0.3) +
    labs(x = "Regular Season Win Percentage (%)",
         y = "Number of Tournament Wins",
         title = "Tournament Wins vs Regular Season Win Percentage") +
    theme(text = element_text(size = 12))
pct_plot
```

We will use KNN regression to predict tourney_wins from seed and reg_wins_pct. This method is appropriate because our response variable is numeric, and KNN makes no assumptions about the shape of the relationship. This matters because our exploratory plots suggest that tournament wins drop sharply across the top seeds and then level off, which is not a straight-line pattern. One limitation is that KNN does not predict well outside the range of the training data and becomes slow with very large data sets. It is also less interpretable than linear regression because it gives no coefficients describing each predictor’s effect. In addition, tourney_wins is right-skewed, so predictions for rare deep tournament runs may be less accurate. To process the data, we will first split the tidied data into a training set (75%) and a testing set (25%) using initial_split, stratified by tourney_wins. Because KNN is distance-based, we will use a recipe to standardize both predictors so that reg_wins_pct, which has a much larger range, does not dominate seed. Using only the training set, we will perform 5-fold cross-validation to choose the number of neighbours K with the lowest RMSPE. This acts as our validation step, so no separate validation set is needed. We will then refit the model with the best K on the whole training set and evaluate it once on the testing set. We will assess performance using RMSPE because it measures the typical prediction error in the same units as the response, which is number of tournament wins, making it easy to interpret.
